```python
#columns to use-title,domain,skills,description
df.copy()
```

```python
main_df=df.drop(['id','organization','location','duration_weeks','eligibility','mode','source'],axis=1)
main_df.copy()
```

```python
main_df["profile_text"] = (
    main_df["title"] + " " +
    main_df["domain"] + " " +
    main_df["skills"].str.replace(";", " ") + " " +
    main_df["description"]
)
```

```python
main_df
```

```python
#TF-IDF text-vector
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(
        stop_words='english',
        max_features=5000
)
tfidf_matrix = vectorizer.fit_transform(main_df["profile_text"])
```

```python
print(tfidf_matrix.shape)
```

```python
#Recommendation Code 
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
def internship_recommendation(user_entry,top_n=5):
    query_vector = vectorizer.transform([user_entry]) #Convert user text into vector
    similarity_scores = cosine_similarity(query_vector, tfidf_matrix)[0]  #Compare user vector with all internships
    top_indices=similarity_scores.argsort()[::-1][:top_n]  #this skip 1.0(itself) from recommendation
    return df.iloc[top_indices]
    
```

```python
print(internship_recommendation('Data Science')[["title", "organization", "domain"]])
```
